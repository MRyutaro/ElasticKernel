```python
import numpy as np
import time

# 2**13の行列（約512MB、float64）
n = 2**13  # 8192x8192の行列
x = np.random.rand(n, n).astype(np.float64)

# 複数の計算を実行して計算時間を延ばす
start_time = time.perf_counter()
for i in range(5):
    y = np.dot(x, x.T)
end_time = time.perf_counter()

elapsed_time = end_time - start_time
print(f"行列サイズ: {n}x{n}")
print(f"計算時間: {elapsed_time:.4f}秒")
```
