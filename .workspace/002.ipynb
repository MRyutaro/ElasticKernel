```python
import time
import tracemalloc

tracemalloc.start()  # メモリ計測開始
start_time = time.perf_counter()

def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

result = fibonacci(40)

end_time = time.perf_counter()
current, peak = tracemalloc.get_traced_memory()
tracemalloc.stop()  # メモリ計測終了

print(f"fibonacci(40) = {result}")
print(f"実行時間: {end_time - start_time:.4f} 秒")
print(f"ピークメモリ使用量: {peak / (1024):.2f} KiB")
```
