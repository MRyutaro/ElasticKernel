```python
import time

start_time = time.perf_counter()

def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

# 計算時間が長くなるように大きな値で実行
result = fibonacci(40)

end_time = time.perf_counter()
elapsed_time = end_time - start_time

print(f"fibonacci(40) = {result}")
print(f"実行時間: {elapsed_time:.4f}秒")
```
